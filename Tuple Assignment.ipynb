# Q42. Find the index of the last occurrence of a specific element in a tuple.

```python
t = (1,2,3,4,3,5,3,6,7)
x = 3
rev_index = t[::-1].index(x)
last_index = len(t) - 1 - rev_index
print(last_index)
```
